## Plotting univariate distributions

The most convenient way to take a quick look at a univariate distribution in seaborn is the distplot() function. By default, this will draw a histogram and fit a kernel density estimate (KDE).



# **distplot**


A distplot in Seaborn provides a combined visualization of a histogram and a kernel density estimation (KDE) plot, allowing for a concise representation of the univariate distribution of a dataset. It's useful for understanding the shape, central tendency, and spread of the data.

```python
import matplotlib.pyplot as plt
import seaborn as sns
bill=data["total_bill"] #univariate

sns.distplot(bill) #kde+hist
plt.show()
```

```python
import matplotlib.pyplot as plt
sns.histplot(bill,color="pink")
plt.show()
```

```python
sns.distplot(bill,hist=False)
```

```python
sns.displot(bill)
```

```python
sns.kdeplot(bill)
```

# **boxplot**



A box plot (also known as a box-and-whisker plot) is a graphical representation of the distribution of a dataset based on five summary statistics:*** minimum, first quartile (Q1), median (Q2), third quartile (Q3), and maximum.***


 It is useful for detecting outliers and comparing distributions between different groups or categories

```python
bill=data["total_bill"]

sns.boxplot(bill) # univariate
```

```python
sns.boxplot(bill,orient='h')#horizondal
```

```python
sns.histplot(bill)
```

```python
sns.kdeplot(bill) #kernel density estimate
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the iris dataset
iris = sns.load_dataset('iris')
iris
```

```python

# Create a scatter plot of sepal length vs. sepal width
sns.lmplot(x='sepal_length', y='sepal_width', data=iris,hue="species",palette="rainbow")
plt.title('Scatter Plot ')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

**palette**

Set1: A qualitative color palette with distinct colors suitable for categorical data.

husl: A perceptually uniform color palette that is suitable for numerical data.

muted: A color palette with muted colors that works well for displaying
multiple data series.

Dark2: A qualitative color palette with darker colors.

```python


# Create a scatter plot of sepal length vs. sepal width
sns.scatterplot(x='sepal_length', y='sepal_width', data=iris, hue='species', palette='Set1', marker='*', s=200)
# Add a title
plt.title('Sepal Length vs. Sepal Width')

# Add labels to the axes
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')

# Add a legend
plt.legend(title='CCY')

# Customize the gridlines
plt.grid(True, linestyle='--')

# Show plot
plt.show()
```

Clustered Distribution: Each species of iris flowers forms a distinct cluster in the plot based on the combination of sepal length and width. This indicates that there are observable differences in sepal dimensions among the different species.

Species Differentiation: The clusters corresponding to different species are visually distinguishable from each other. This suggests that sepal length and width can potentially be used as features to discriminate between iris species.


## Joint Plot:

A joint plot displays the relationship between two variables along with the individual distributions of each variable.
Seaborn's jointplot() function can be used to create joint plots.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the iris dataset
iris = sns.load_dataset('iris')

# Create a joint plot of sepal length vs. sepal width
sns.jointplot(x='sepal_length', y='sepal_width', data=iris)
plt.title('Joint Plot of Sepal Length vs. Sepal Width')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the iris dataset
iris = sns.load_dataset('iris')

# Create a joint plot of sepal length vs. sepal width
sns.jointplot(x='sepal_length', y='sepal_width', data=iris,hue="species")
plt.title('Joint Plot of Sepal Length vs. Sepal Width')
plt.show()
```

Seaborn's barplot() function is suitable for showing an aggregate statistic (e.g., mean, median) of a numerical variable across different categories.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset('tips')

# Create a bar plot of average tip amount by day
sns.barplot(x='day', y='tip', data=tips) # 1 cat , 1 nu
plt.title('Average Tip Amount by Day')
plt.xlabel('Day')
plt.ylabel('Average Tip Amount')
plt.show()
```

## **count plot**

shows the frequency of occurrences of each category in a categorical variable.
Seaborn's countplot() function is useful for visualizing the distribution of categorical data.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset('tips')

# Create a count plot of the number of observations by day
sns.countplot(x='sex', data=tips) # univariate
plt.title('Number of Observations by Day')
plt.xlabel('Day')
plt.ylabel('Count')
plt.show()
```

**Box Plot:**

A box plot can be used to visualize the distribution of a numerical variable across different categories.
Seaborn's boxplot() function is suitable for comparing distributions of numerical variables across categories.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset('tips')

# Create a box plot of total bill by day
sns.boxplot(x='day', y='total_bill', data=tips) #1 cat, 1 nu
plt.title('Distribution of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

**Violin Plot:**


A violin plot combines the features of a box plot and a kernel density estimation plot to visualize the distribution of a numerical variable across different categories.


Seaborn's violinplot() function is useful for comparing the distribution of numerical variables across categories.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset('tips')

# Create a violin plot of total bill by day
sns.violinplot(x='day', y='total_bill', data=tips) #1 cat, 1 nu
plt.title('Distribution of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

**Strip Plot:**

A strip plot is a one-dimensional scatter plot where the categorical variable is plotted on the x-axis, and the continuous variable is plotted on the y-axis. Each data point is represented as a single point along the categorical axis.
Seaborn's stripplot() function is used to create strip plots.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset('tips')

# Create a strip plot of total bill by day
sns.stripplot(x='day', y='total_bill', data=tips,hue="sex") #1cat, 1 nu
plt.title('Strip Plot of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

**Swarm Plot:**

A swarm plot is similar to a strip plot, but it spreads out the points to avoid overlap, giving a better visualization of the distribution of the data.
Seaborn's swarmplot() function is used to create swarm plots.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset('tips')

# Create a swarm plot of total bill by day
sns.swarmplot(x='day', y='total_bill', data=tips,hue="sex")
plt.title('Swarm Plot of Total Bill by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
```

**A factorplot (formerly known as catplot)**

A factorplot (formerly known as catplot) in Seaborn is a versatile function that allows you to create different types of plots by specifying the kind parameter. It's especially useful for visualizing the distribution of a variable across multiple categories, as well as for comparing relationships between variables within different subsets of the data. Here's how you can use factorplot:

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset("tips")

# Create a catplot of total bill by day
sns.catplot(x="day", y="total_bill", data=tips, kind="box") # 1 cat, 1 numerical

# Add titles and labels
plt.title("Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")



# Show the plot
plt.show()
```

```python
# Load the tips dataset
tips = sns.load_dataset("tips")

# Create a catplot of total bill by day
sns.catplot(x="time", y="total_bill",hue="smoker",col="day", data=tips, kind="bar") #kind= 'strip', 'swarm', 'box', 'boxen', 'violin', 'bar', 'count', and 'point'

# Add titles and labels
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")
# Set the super title and adjust its vertical position
# Add a title and adjust its vertical position
plt.suptitle("Total Bill by Day, Separated by Time and Smoker Status", y=1.05)

# Show the plot
plt.show()
```

## Function to draw linear regression models

lmplot() is one of the most widely used function to quickly plot the Linear Relationship b/w 2 variables


```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset("tips")

# Create an lmplot of total bill vs. tip with hue based on smoker status
sns.lmplot(x="total_bill", y="tip", data=tips) #2 numerical

# Add titles and labels
plt.title("Linear Regression Plot of Total Bill vs. Tip")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")

# Show the plot
plt.show()
```
